Task:Read age and check voting eligibility.


```python
age = int(input("Enter the age :"))
print("Enter the age of the person:",age)

if age >= 18:
    print("Eligible for the vote.")
else:
    print("Not Eligible:")
```

Task: Read marks and display the grade

```python
marks = int(input("Enter the marks of the student:"))
print("Enter the marks of the student",marks)
if marks >= 90:
    print("Grade A")
elif marks >=80:
    print("Grade B")
elif marks >=60:
    print("Grade C")
elif marks >=40:
    print("Grade D")
else:
    print("Fail")
```
