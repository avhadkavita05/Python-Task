## Different Methods in list

#### 1) append()-----> It is used to add one element at the end of the list

```python
a = [10, 20, 30,"A","Ball",13.5]
a.append(40)

print(a)
```

#### 2) extend()----> It is used to adds multiple elements in the list

```python
s=['s','apple','print',12,34]
s.extend([12,"Sneha",True])
print(s)
```

#### 3)insert()-----> It is used to add element at a particular index in the list

```python
a = [10, 20, 30,"A","Ball",13.5]
a.insert(1,23)
print(a)
```

#### 4) remove()------> It is used to removes a particular value from list.

```python
a = [10, 20, 30,"A","Ball",13.5]
a.remove("Ball")
print(a)
```

#### 5) pop()---> It is used to remove an element using its index.

```python
s=[10,23,4,56,78,34]
s.pop(4)
print(s)
```

#### 6) sort()------> Used to sort the elements in ascending order.

```python
b=[40,56,0,68]
b.sort()
print(b)
```

#### 7) reverse()-----> It is used to reverse the list.

```python
a = [10, 20, 30,"A","Ball",13.5]
a.reverse()
print(a)
```

#### 8) count()------> used to count how many times a value occurs.

```python
a = [10, 20, 10, 30, 10]

print(a.count(10))
```

#### 9) index()-----> used to returns of an position of an element.

```python
a=["kavita",30,"Rahul",21,45,True]
print(a.index("Rahul"))
```

#### 10)clear()-----> used to removes everything from list.

```python
a = [10, 20, 30]

a.clear()

print(a)
```

#### copy()------> The copy() method creates a new list containing the same elements as the original list.

#### Syntax----> new_list = old_list.copy()

```python
a = [10, 20, 30, 40]

b = a.copy()

print("Original list =", a)
print("Copied list =", b)
```
