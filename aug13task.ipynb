# Task = Perform any 5 string operation on your name
### Positive index
### Negative index 
### positive slice
### Negative slice

```python
Name = "AvhadKavita"
print(Name)
```

```python
#Positive Index
print(Name[1])
print(Name[0])
print(Name[5])
print(Name[10]) 
A = Name[0]+Name[1]+Name[2]+Name[3]+Name[4]
print(A)
```

```python
#Negative index
print(Name[-3]) 
print(Name[-10]) 
print(Name[-6]) 
print(Name[-11])
print(Name[-7]) 
```

```python
# Positive Slicing
print(Name[0:5])
print(Name[2:5])
print(Name[5:8])
print(Name[0:8])
print(Name[0:10])
```

```python
#Negative slicing
print(Name[-6:-1])
print(Name[-8:-3])
print(Name[-9:-2])
print(Name[-4:])
print(Name[:-6])
```
