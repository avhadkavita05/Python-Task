```python
#Assignment Operators = Assignment operators are used to assign a value to a variable.

##The basic assignment operator is =.
```

```python
a = 10
print(a)
```

```python
a = 10

a += 5
print(a)       

a -= 3
print(a)       

a *= 2
print(a)       

a //= 4
print(a)       

a **= 2
print(a)       
```

```python
#Compound Assignment Operators
#Compound assignment operators combine an arithmetic operation + assignment into one operator.
```

```python
a = 20

a += 5
print(a)       

a -= 10
print(a)       

a *= 2
print(a)       

a /= 3
print(a)       

a %= 3
print(a)       
```

```python
#Bitwise Operators
#Bitwise operators perform operations on the binary bits (0 and 1) of numbers.
```

```python
#Bitwise AND &
a = 10
b = 6

print(a & b)
```

```python
#Bitwise OR |
a = 10
b = 6

print(a | b)
```

```python
#Bitwise XOR ^
a = 10
b = 6

print(a ^ b)
```

```python
#Left Shift <<
a = 5

print(a << 1)
print(a << 2)
```

```python
#Right Shift >>
a = 20

print(a >> 1)
print(a >> 2)
```

```python
#Bitwise NOT ~ [FORMULA =~n = -(n + 1)]
a = 10

print(~a)
```
